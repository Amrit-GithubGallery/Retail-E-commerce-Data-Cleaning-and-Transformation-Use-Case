# DATA CLEANING

#### Scenario 1: Trim whitespace and standardize IDs

```python
# CLEANING SCENARIO 1:
# Remove unnecessary whitespace from order IDs,
# customer IDs, product IDs and seller IDs.

orders["order_id"] = orders["order_id"].str.strip()
orders["customer_id"] = orders["customer_id"].str.strip()

customers["customer_id"] = customers["customer_id"].str.strip()
customers["customer_unique_id"] = customers["customer_unique_id"].str.strip()

order_items["order_id"] = order_items["order_id"].str.strip()
order_items["product_id"] = order_items["product_id"].str.strip()
order_items["seller_id"] = order_items["seller_id"].str.strip()

payments["order_id"] = payments["order_id"].str.strip()

products["product_id"] = products["product_id"].str.strip()

sellers["seller_id"] = sellers["seller_id"].str.strip()

order_reviews["order_id"] = order_reviews["order_id"].str.strip()

display(orders.head())
print(orders.shape)
```

### Scenario 2: Parse timestamps and convert to UTC

```python
# CLEANING SCENARIO 2:
# Convert order date columns into datetime format.
# UTC conversion assumes the source timestamps represent UTC.

orders["order_purchase_timestamp"] = pd.to_datetime(
    orders["order_purchase_timestamp"], utc=True, errors="coerce"
)

orders["order_approved_at"] = pd.to_datetime(
    orders["order_approved_at"], utc=True, errors="coerce"
)

orders["order_delivered_carrier_date"] = pd.to_datetime(
    orders["order_delivered_carrier_date"], utc=True, errors="coerce"
)

orders["order_delivered_customer_date"] = pd.to_datetime(
    orders["order_delivered_customer_date"], utc=True, errors="coerce"
)

orders["order_estimated_delivery_date"] = pd.to_datetime(
    orders["order_estimated_delivery_date"], utc=True, errors="coerce"
)

order_items["shipping_limit_date"] = pd.to_datetime(
    order_items["shipping_limit_date"], utc=True, errors="coerce"
)

order_reviews["review_creation_date"] = pd.to_datetime(
    order_reviews["review_creation_date"], utc=True, errors="coerce"
)

order_reviews["review_answer_timestamp"] = pd.to_datetime(
    order_reviews["review_answer_timestamp"], utc=True, errors="coerce"
)

display(orders.head())
print(orders.dtypes)
```

### Scenario 3: Standardize product category names

```python
# CLEANING SCENARIO 3:
# Standardize product category names.
# Use the category translation dataset to get English category names.
# Olist does not provide product names, SKU codes or brand names.

products["product_category_name"] = (
    products["product_category_name"]
    .astype("string")
    .str.strip()
    .str.lower()
)

category_translation["product_category_name"] = (
    category_translation["product_category_name"]
    .astype("string")
    .str.strip()
    .str.lower()
)

products = products.merge(
    category_translation,
    on="product_category_name",
    how="left"
)

products["product_category_name_english"] = (
    products["product_category_name_english"]
    .fillna(products["product_category_name"])
)

display(products.head())
print(products.shape)
```

### Scenario 4: Normalize quantities and package units

```python
# CLEANING SCENARIO 4:
# Olist does not contain an explicit quantity or package-unit column.
# Each order-item row is treated as one assumed unit.
# This is an assumption, not a measured quantity.

order_items["assumed_line_units"] = 1

display(order_items.head())
print(order_items.shape)
```

### Scenario 5: Convert prices and payment values into numeric format

```python
# CLEANING SCENARIO 5:
# Convert price, freight and payment values into numeric format.
# Discounts, taxes and refunds are not available in standard Olist files.

order_items["price"] = pd.to_numeric(
    order_items["price"], errors="coerce"
)

order_items["freight_value"] = pd.to_numeric(
    order_items["freight_value"], errors="coerce"
)

payments["payment_value"] = pd.to_numeric(
    payments["payment_value"], errors="coerce"
)

display(order_items.head())
display(payments.head())
```

### Scenario 6: Standardize currencies

```python
# CLEANING SCENARIO 6:
# Olist monetary values are generally represented in Brazilian Real (BRL).
# No currency conversion is performed because no currency or FX-rate table
# is included in the standard dataset.

print("Currency conversion cannot be performed without an FX-rate dataset.")
```

### Scenario 7: Standardize geography

```python
# CLEANING SCENARIO 7:
# Standardize customer and seller city/state names.
# Keep postal codes as strings to preserve leading zeroes.

customers["customer_state"] = (
    customers["customer_state"].str.strip().str.upper()
)

customers["customer_city"] = (
    customers["customer_city"].str.strip().str.lower()
)

sellers["seller_state"] = (
    sellers["seller_state"].str.strip().str.upper()
)

sellers["seller_city"] = (
    sellers["seller_city"].str.strip().str.lower()
)

customers["customer_zip_code_prefix"] = (
    customers["customer_zip_code_prefix"].astype(str).str.zfill(5)
)

sellers["seller_zip_code_prefix"] = (
    sellers["seller_zip_code_prefix"].astype(str).str.zfill(5)
)

display(customers.head())
display(sellers.head())
```

### Scenario 8: Detect and remove duplicate order lines

```python
# CLEANING SCENARIO 8:
# Identify duplicate order lines using order_id and order_item_id.
# Keep the first record when duplicates are found.
# Olist does not provide event-version information.

print("Duplicate rows:", order_items.duplicated(
    subset=["order_id", "order_item_id"]
).sum())

order_items = order_items.drop_duplicates(
    subset=["order_id", "order_item_id"],
    keep="first"
)
display(order_items.head())
print(order_items.shape)
```

### Scenario 9: Validate price and freight values

```python
# CLEANING SCENARIO 9:
# Identify invalid negative or missing price and freight values.
# Remove invalid rows before sales analysis.

print("Invalid price rows:",
      order_items[order_items["price"].isna() |
                  (order_items["price"] < 0)].shape)

print("Invalid freight rows:",
      order_items[order_items["freight_value"].isna() |
                  (order_items["freight_value"] < 0)].shape)

order_items = order_items[
    (order_items["price"].notna()) &
    (order_items["price"] >= 0) &
    (order_items["freight_value"].notna()) &
    (order_items["freight_value"] >= 0)
]

order_items["gross_line_value"] = (
    order_items["price"] * order_items["assumed_line_units"]
)

display(order_items.head())
```

### Scenario 10: Normalize payment methods

```python
# CLEANING SCENARIO 10:
# Standardize payment method labels into lowercase format.

payments["payment_type"] = (
    payments["payment_type"].str.strip().str.lower()
)

payments["payment_type"] = payments["payment_type"].replace({
    "credit card": "credit_card",
    "debit card": "debit_card"
})

display(payments.head())
```

### Scenario 11: Match customer records with master data

```python
# CLEANING SCENARIO 11:
# Check whether every customer_id in orders exists in customers.
# Identify unmatched customer IDs.

unmatched_customers = orders[
    ~orders["customer_id"].isin(customers["customer_id"])
]

print("Unmatched customer records:", unmatched_customers.shape)


display(unmatched_customers.head())
```

### Scenario 12: Normalize promotion codes and statuses

```python
# CLEANING SCENARIO 12:
# Olist does not contain promotion codes, coupon codes or campaign statuses.
# This scenario cannot be performed with the available dataset.

print("Promotion code data is not available.")
```

### Scenario 13: Standardize return and refund reasons

```python
# CLEANING SCENARIO 13:
# Olist does not contain a return/refund dataset or reason codes.
# Return and refund standardization cannot be performed.

print("Return and refund reason data is not available.")
```

### Scenario 14: Standardize shipment and delivery status

```python
# CLEANING SCENARIO 14:
# Standardize order status values.
# Delivery timestamps are available, but tracking IDs and carrier names
# are not included in the standard Olist dataset.

orders["order_status"] = (
    orders["order_status"].str.strip().str.lower()
)

display(orders[["order_id", "order_status"]].head())
```

### Scenario 15: Mask customer identifiers

```python
# CLEANING SCENARIO 15:
# Create tokenized customer identifiers for analyst extracts.
# Remove original customer IDs from the privacy-safe dataset.

customer_token_map = pd.DataFrame({
    "customer_id": customers["customer_id"].dropna().drop_duplicates()
})

customer_token_map["customer_token"] = [
    "CUST_" + str(i).zfill(6)
    for i in range(1, len(customer_token_map) + 1)
]

customers_safe = customers.merge(
    customer_token_map,
    on="customer_id",
    how="left"
)

customers_safe = customers_safe.drop(
    columns=["customer_id", "customer_unique_id"]
)

# Display privacy-safe dataset
display(customers_safe.head())
```

### Scenario 16: Identify suspicious payment values

```python
# CLEANING SCENARIO 16:
# Identify unusually high payment values using the IQR method.
# Failed-payment and refund-ratio analysis is unavailable because
# payment status and refund columns are missing.

payment_order_totals = payments.groupby(
    "order_id"
)["payment_value"].sum().reset_index()

Q1 = payment_order_totals["payment_value"].quantile(0.25)

Q3 = payment_order_totals["payment_value"].quantile(0.75)

IQR = Q3 - Q1

payment_order_totals["high_payment_flag"] = (
    payment_order_totals["payment_value"] > Q3 + 1.5 * IQR
)

display(payment_order_totals.head())
```

### Scenario 17: Validate product and seller keys

```python
# CLEANING SCENARIO 17:
# Check whether product IDs and seller IDs in order_items
# exist in their respective master datasets.
# Inventory and supplier master data are not available.

invalid_products = order_items[
    ~order_items["product_id"].isin(products["product_id"])
]

invalid_sellers = order_items[
    ~order_items["seller_id"].isin(sellers["seller_id"])
]

print("Invalid product IDs:", invalid_products.shape)

print("Invalid seller IDs:", invalid_sellers.shape)

display(invalid_products.head())
display(invalid_sellers.head())
```

### Scenario 18: Separate cancelled and fulfilled orders

```python
# CLEANING SCENARIO 18:
# Separate delivered orders and cancelled orders.
# Olist does not provide test-order or fraud-review flags.

fulfilled_orders = orders[
    orders["order_status"] == "delivered"
]

cancelled_orders = orders[
    orders["order_status"] == "canceled"
]

print("Fulfilled orders:", fulfilled_orders.shape)

print("Cancelled orders:", cancelled_orders.shape)

display(fulfilled_orders.head())
display(cancelled_orders.head())
```

### Scenario 19: Handle missing values

```python
# CLEANING SCENARIO 19:
# Identify missing values and their percentages.
# Do not fill missing customer IDs, discounts or unknown values
# with invented information.

missing_values = orders.isnull().sum()

missing_percentage = (
    orders.isnull().mean() * 100
).round(2)

missing_summary = pd.DataFrame({
    "Missing Values": missing_values,
    "Missing Percentage": missing_percentage
})

display(missing_summary)
```

### Scenario 20: Detect out-of-order events

```python
# CLEANING SCENARIO 20:
# Check whether order timestamps occur in a logical sequence.
# Olist does not contain ingestion timestamps, so late-arriving
# updates cannot be identified directly.

orders["approval_before_purchase"] = (
    orders["order_approved_at"] <
    orders["order_purchase_timestamp"]
)

orders["carrier_before_approval"] = (
    orders["order_delivered_carrier_date"] <
    orders["order_approved_at"]
)

orders["delivery_before_carrier"] = (
    orders["order_delivered_customer_date"] <
    orders["order_delivered_carrier_date"]
)

display(orders.head())
```

# Prepare the main sales dataset

```python
# ============================================================
# PREPARE MAIN SALES DATASET
# Merge orders, products, customers and sellers with order_items.
# This creates one combined dataset for business analysis.
# ============================================================

sales = order_items.merge(
    orders,
    on="order_id",
    how="left"
)

sales = sales.merge(
    products[["product_id", "product_category_name_english"]],
    on="product_id",
    how="left"
)

sales = sales.merge(
    customers[[
        "customer_id",
        "customer_unique_id",
        "customer_city",
        "customer_state"
    ]],
    on="customer_id",
    how="left"
)

sales = sales.merge(
    sellers[[
        "seller_id",
        "seller_city",
        "seller_state"
    ]],
    on="seller_id",
    how="left"
)

# Create date and month columns for sales analysis.
sales["purchase_date"] = sales["order_purchase_timestamp"].dt.date

sales["purchase_month"] = (
    sales["order_purchase_timestamp"].dt.to_period("M").astype(str)
)

sales["gross_sales"] = sales["gross_line_value"]

sales["shipping_value"] = sales["freight_value"]

display(sales.head())

print("Sales dataset shape:", sales.shape)
```

# DATA TRANSFORMATION

### Scenario 1: Daily and monthly gross sales

```python
# TRANSFORMATION SCENARIO 1:
# Calculate daily and monthly gross sales by product category.
# Discounts, taxes, refunds and sales channels are not available.
# Therefore, this is a gross-sales summary, not a full net-sales report.

daily_sales = sales.groupby(
    ["purchase_date", "product_category_name_english"]
).agg(
    gross_sales=("gross_sales", "sum"),
    shipping_value=("shipping_value", "sum"),
    units=("assumed_line_units", "sum"),
    orders=("order_id", "nunique")
).reset_index()

monthly_sales = sales.groupby(
    ["purchase_month", "product_category_name_english"]
).agg(
    gross_sales=("gross_sales", "sum"),
    shipping_value=("shipping_value", "sum"),
    units=("assumed_line_units", "sum"),
    orders=("order_id", "nunique")
).reset_index()

# Display results
display(daily_sales.head())
display(monthly_sales.head())
```

### Scenario 2: Average order value and items per order

```python
# TRANSFORMATION SCENARIO 2:
# Calculate average order value and average number of item rows per order.
# Revenue per visitor cannot be calculated because visitor data is absent.

order_summary = sales.groupby("order_id").agg(
    total_items=("order_item_id", "count"),
    total_sales=("gross_sales", "sum")
).reset_index()

average_order_value = order_summary["total_sales"].mean()

average_items = order_summary["total_items"].mean()

print("Average Order Value:", round(average_order_value, 2))

print("Average Items Per Order:", round(average_items, 2))

display(order_summary.head())
```

### Scenario 3: Customer cohort analysis

```python
# TRANSFORMATION SCENARIO 3:
# Create customer cohorts using the month of their first purchase.
# Calculate repeat-purchase activity by month since first purchase.

customer_orders = sales[[
    "customer_unique_id",
    "order_id",
    "order_purchase_timestamp"
]].dropna().drop_duplicates()

first_purchase = customer_orders.groupby(
    "customer_unique_id"
)["order_purchase_timestamp"].min()

customer_orders["first_purchase"] = (
    customer_orders["customer_unique_id"].map(first_purchase)
)

customer_orders["cohort_month"] = (
    customer_orders["first_purchase"].dt.to_period("M").astype(str)
)

customer_orders["order_month"] = (
    customer_orders["order_purchase_timestamp"].dt.to_period("M").astype(str)
)

customer_orders["months_since_first_purchase"] = (
    (customer_orders["order_purchase_timestamp"].dt.year -
     customer_orders["first_purchase"].dt.year) * 12
    +
    customer_orders["order_purchase_timestamp"].dt.month -
    customer_orders["first_purchase"].dt.month
)

cohort_analysis = customer_orders.groupby(
    ["cohort_month", "months_since_first_purchase"]
)["customer_unique_id"].nunique().reset_index(name="active_customers")

cohort_sizes = customer_orders.groupby(
    "cohort_month"
)["customer_unique_id"].nunique()

cohort_analysis["cohort_customers"] = (
    cohort_analysis["cohort_month"].map(cohort_sizes)
)

cohort_analysis["repeat_rate"] = (
    cohort_analysis["active_customers"] /
    cohort_analysis["cohort_customers"] * 100
).round(2)

display(cohort_analysis.head())
```

### Scenario 8: Delivery performance analysis

```python
# TRANSFORMATION SCENARIO 8:
# Calculate fulfilment time and delivery performance.
# Compare actual delivery date with estimated delivery date.
# Carrier performance cannot be calculated because carrier names are absent.

delivery = orders.copy()

delivery["fulfilment_days"] = (
    delivery["order_delivered_customer_date"] -
    delivery["order_purchase_timestamp"]
).dt.total_seconds() / 86400

delivery["delivery_difference"] = (
    delivery["order_delivered_customer_date"] -
    delivery["order_estimated_delivery_date"]
).dt.total_seconds() / 86400

delivery["delivery_category"] = np.select(
    [
        delivery["order_delivered_customer_date"].isna(),
        delivery["delivery_difference"] < 0,
        delivery["delivery_difference"] > 0
    ],
    [
        "Not Delivered",
        "Early",
        "Late"
    ],
    default="On-Time"
)

delivery_summary = delivery.groupby(
    "delivery_category"
).agg(
    total_orders=("order_id", "count"),
    average_delivery_days=("fulfilment_days", "mean")
).reset_index()

display(delivery_summary)
```
